# Scraping Dataset

```python
# Melakukan scraping ulasan aplikasi Telkomsel
telkomsel_reviews = reviews_all(
    'com.telkomsel.telkomselcm',  # ID aplikasi Telkomsel
    lang='id',                    # Bahasa ulasan (Bahasa Indonesia)
    country='id',                 # Negara (Indonesia)
    sort=Sort.MOST_RELEVANT,             # Urutan ulasan (Terbaru)
    count=10001                    # Jumlah maksimum ulasan
)

# Mengonversi hasil scraping ke dalam DataFrame
reviews_df = pd.DataFrame(telkomsel_reviews)

# Memilih kolom yang relevan saja
reviews_df = reviews_df[['content', 'score', 'at', 'userName']]

# Mengubah nama kolom agar lebih deskriptif
reviews_df.columns = ['review', 'rating', 'date', 'user']

# Menampilkan 5 data teratas
print(reviews_df.head())
```
